```python
# Smart Bus Reservation System

buses = {
    1: {
        "name": "Indore - Bhopal Express",
        "from": "Indore",
        "to": "Bhopal",
        "time": "08:00 AM",
        "fare": 350,
        "seats": {}
    },
    2: {
        "name": "Indore - Ujjain Bus",
        "from": "Indore",
        "to": "Ujjain",
        "time": "10:30 AM",
        "fare": 200,
        "seats": {}
    },
    3: {
        "name": "Indore - Delhi Travels",
        "from": "Indore",
        "to": "Delhi",
        "time": "07:00 PM",
        "fare": 900,
        "seats": {}
    }
}

TOTAL_SEATS = 20


def show_buses():
    print("\n========== AVAILABLE BUSES ==========")

    for bus_id, bus in buses.items():
        available = TOTAL_SEATS - len(bus["seats"])

        print(f"\nBus ID: {bus_id}")
        print("Bus Name:", bus["name"])
        print("Route:", bus["from"], "to", bus["to"])
        print("Departure:", bus["time"])
        print("Fare: ₹", bus["fare"])
        print("Available Seats:", available)


def show_seats():
    show_buses()

    bus_id = int(input("\nEnter Bus ID: "))

    if bus_id not in buses:
        print("Invalid Bus ID!")
        return

    print("\n========== SEAT STATUS ==========")

    for seat in range(1, TOTAL_SEATS + 1):

        if seat in buses[bus_id]["seats"]:
            print(f"Seat {seat}: BOOKED")
        else:
            print(f"Seat {seat}: AVAILABLE")


def book_ticket():
    show_buses()

    bus_id = int(input("\nEnter Bus ID: "))

    if bus_id not in buses:
        print("Invalid Bus ID!")
        return

    bus = buses[bus_id]

    if len(bus["seats"]) >= TOTAL_SEATS:
        print("Sorry! All seats are booked.")
        return

    print("\nAvailable seats:")

    for seat in range(1, TOTAL_SEATS + 1):
        if seat not in bus["seats"]:
            print(seat, end=" ")

    print()

    seat_number = int(input("\nEnter seat number: "))

    if seat_number < 1 or seat_number > TOTAL_SEATS:
        print("Invalid seat number!")
        return

    if seat_number in bus["seats"]:
        print("Sorry! This seat is already booked.")
        return

    name = input("Enter passenger name: ")
    age = int(input("Enter passenger age: "))
    phone = input("Enter phone number: ")

    bus["seats"][seat_number] = {
        "name": name,
        "age": age,
        "phone": phone
    }

    print("\n================================")
    print("       TICKET BOOKED!")
    print("================================")
    print("Passenger:", name)
    print("Bus:", bus["name"])
    print("Route:", bus["from"], "to", bus["to"])
    print("Departure:", bus["time"])
    print("Seat Number:", seat_number)
    print("Fare: ₹", bus["fare"])
    print("================================")


def cancel_ticket():
    show_buses()

    bus_id = int(input("\nEnter Bus ID: "))

    if bus_id not in buses:
        print("Invalid Bus ID!")
        return

    bus = buses[bus_id]

    seat_number = int(input("Enter seat number to cancel: "))

    if seat_number not in bus["seats"]:
        print("No booking found for this seat.")
        return

    passenger = bus["seats"][seat_number]

    print("\nPassenger:", passenger["name"])
    confirm = input("Are you sure you want to cancel? (yes/no): ")

    if confirm.lower() == "yes":
        del bus["seats"][seat_number]
        print("Ticket cancelled successfully.")
    else:
        print("Cancellation stopped.")


def view_bookings():
    print("\n========== ALL BOOKINGS ==========")

    found = False

    for bus_id, bus in buses.items():

        if len(bus["seats"]) > 0:
            found = True

            print("\nBus:", bus["name"])

            for seat, passenger in bus["seats"].items():
                print("----------------------------")
                print("Seat:", seat)
                print("Name:", passenger["name"])
                print("Age:", passenger["age"])
                print("Phone:", passenger["phone"])

    if not found:
        print("No bookings available.")


# Main Program

while True:

    print("\n======================================")
    print("     SMART BUS RESERVATION SYSTEM")
    print("======================================")
    print("1. View Available Buses")
    print("2. View Seat Status")
    print("3. Book Ticket")
    print("4. Cancel Ticket")
    print("5. View All Bookings")
    print("6. Exit")
    print("======================================")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_buses()

    elif choice == "2":
        show_seats()

    elif choice == "3":
        book_ticket()

    elif choice == "4":
        cancel_ticket()

    elif choice == "5":
        view_bookings()

    elif choice == "6":
        print("\nThank you for using Smart Bus Reservation System!")
        print("Have a safe journey! 🚌")
        break

    else:
        print("Invalid choice! Please try again.")
```
