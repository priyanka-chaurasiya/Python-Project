```python
# Personal Budget Manager

budget = 0
income = 0
expenses = []

while True:
    print("\n===== Personal Budget Manager =====")
    print("1. Set Budget")
    print("2. Add Income")
    print("3. Add Expense")
    print("4. View Expenses")
    print("5. View Summary")
    print("6. Exit")

    choice = input("Enter your choice: ")

    # Set Budget
    if choice == "1":
        budget = float(input("Enter your monthly budget: ₹"))
        print("Budget set successfully!")

    # Add Income
    elif choice == "2":
        amount = float(input("Enter income amount: ₹"))
        income += amount
        print("Income added successfully!")

    # Add Expense
    elif choice == "3":
        name = input("Enter expense name: ")
        amount = float(input("Enter expense amount: ₹"))

        expenses.append({
            "name": name,
            "amount": amount
        })

        print("Expense added successfully!")

    # View Expenses
    elif choice == "4":
        if len(expenses) == 0:
            print("No expenses found.")
        else:
            print("\n----- Expenses -----")

            for i, expense in enumerate(expenses, start=1):
                print(i, expense["name"], "₹", expense["amount"])

    # View Summary
    elif choice == "5":
        total_expense = sum(expense["amount"] for expense in expenses)
        balance = budget + income - total_expense

        print("\n----- Budget Summary -----")
        print("Budget: ₹", budget)
        print("Income: ₹", income)
        print("Total Expense: ₹", total_expense)
        print("Remaining Balance: ₹", balance)

        if balance < 0:
            print("Warning: You are over your budget!")

        elif balance == 0:
            print("Your budget is completely used.")

        else:
            print("You are within your budget.")

    # Exit
    elif choice == "6":
        print("Thank you for using Personal Budget Manager!")
        break

    else:
        print("Invalid choice! Please try again.")
```
