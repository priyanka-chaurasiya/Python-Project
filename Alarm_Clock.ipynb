```python
import time
from datetime import datetime

alarm_time = input("Enter alarm time (HH:MM): ")

print("Alarm set for", alarm_time)

while True:
    current_time = datetime.now().strftime("%H:%M")

    if current_time == alarm_time:
        print("Wake Up! Alarm is ringing!")
        break

    time.sleep(1)
```
