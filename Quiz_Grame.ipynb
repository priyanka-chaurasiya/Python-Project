```python
# Quiz Game

score = 0

print("===== Welcome to the Quiz Game =====")

# Question 1
print("\n1. What is the capital of India?")
print("A. Mumbai")
print("B. New Delhi")
print("C. Kolkata")
print("D. Chennai")

answer = input("Enter your answer: ")

if answer.upper() == "B":
    print("Correct!")
    score += 1
else:
    print("Wrong! The correct answer is B.")

# Question 2
print("\n2. Which language is used for Python programming?")
print("A. Python")
print("B. HTML")
print("C. CSS")
print("D. SQL")

answer = input("Enter your answer: ")

if answer.upper() == "A":
    print("Correct!")
    score += 1
else:
    print("Wrong! The correct answer is A.")

# Question 3
print("\n3. What is the extension of a Python file?")
print("A. .java")
print("B. .html")
print("C. .py")
print("D. .cpp")

answer = input("Enter your answer: ")

if answer.upper() == "C":
    print("Correct!")
    score += 1
else:
    print("Wrong! The correct answer is C.")

# Final Score
print("\n===== Quiz Completed =====")
print("Your Score:", score, "/ 3")

if score == 3:
    print("Excellent!")
elif score == 2:
    print("Good Job!")
elif score == 1:
    print("Keep Practicing!")
else:
    print("Try Again!")
```
